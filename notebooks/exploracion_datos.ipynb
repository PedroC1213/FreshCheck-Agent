# Exploración de datos (FruitVision, fresca vs. dañada)

Ejecutar desde la raíz del repositorio, después de `python -m src.run_baseline`.

```python
import pandas as pd, matplotlib.pyplot as plt, json
m = pd.read_csv('data/manifest.csv'); s = pd.read_csv('data/splits.csv').merge(m, on='id')
print(len(m), 'imágenes')
display(m.pivot_table(index='fruta', columns='clase', values='id', aggfunc='count'))
display(s.pivot_table(index='clase', columns='particion', values='id', aggfunc='count')[['train','val','test']])
```

```python
m['res'] = m['ancho'].astype(str) + 'x' + m['alto'].astype(str)
print('resoluciones distintas:', m['res'].nunique()); print(m['res'].value_counts().head(5))
fig, ax = plt.subplots(1, 2, figsize=(8, 3))
m['ancho'].hist(ax=ax[0], bins=30); ax[0].set_title('Ancho (px)')
m['alto'].hist(ax=ax[1], bins=30); ax[1].set_title('Alto (px)'); plt.tight_layout(); plt.show()
```

```python
from PIL import Image
fig, axs = plt.subplots(2, 5, figsize=(14, 6))
for j, f in enumerate(['manzana','banano','mango','naranja','uva']):
    for i, c in enumerate(['fresca','danada']):
        fila = m[(m.fruta == f) & (m.clase == c)].iloc[0]
        im = Image.open('data/raw/' + fila['id']); im.draft('RGB', (400, 400))
        axs[i, j].imshow(im.convert('RGB')); axs[i, j].set_title(f'{f} - {c}'); axs[i, j].axis('off')
plt.tight_layout(); plt.show()
```
